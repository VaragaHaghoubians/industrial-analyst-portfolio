# Demand

One row is one SKU in one week. Run `src/main.py` first so the CSVs exist.

**Synthetic inventory sample. Not a purchasing decision.**


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

weekly = pd.read_csv(ROOT / "data/raw/weekly_demand.csv", parse_dates=["week_start"])
master = pd.read_csv(ROOT / "data/raw/sku_master.csv")
print(len(weekly), "weekly rows,", master["sku"].nunique(), "SKUs")
weekly.groupby("sku")["demand_units"].agg(["sum", "mean", "std"]).round(1)
```
