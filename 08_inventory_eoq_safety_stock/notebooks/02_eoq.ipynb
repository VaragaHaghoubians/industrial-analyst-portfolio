# Economic order quantity

EOQ is the square root of `2 x annual demand x order cost / holding cost`. At that quantity, ordering cost and cycle-stock holding cost meet.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from inventory import build_policy

weekly = pd.read_csv(ROOT / "data/raw/weekly_demand.csv", parse_dates=["week_start"])
master = pd.read_csv(ROOT / "data/raw/sku_master.csv")
policy = build_policy(weekly, master)
policy[["sku", "annual_demand", "eoq", "order_qty", "ordering_cost", "holding_cost", "total_relevant_cost"]].round(1)
```
