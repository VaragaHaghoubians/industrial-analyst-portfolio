# Safety stock and the reorder point

Safety stock grows with the square root of lead time. The reorder point adds that buffer to expected demand during the lead time.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from inventory import build_policy

weekly = pd.read_csv(ROOT / "data/raw/weekly_demand.csv", parse_dates=["week_start"])
master = pd.read_csv(ROOT / "data/raw/sku_master.csv")
policy = build_policy(weekly, master)
policy[["sku", "service_level", "lead_time_weeks", "safety_stock", "reorder_point", "on_hand", "action"]].round(1)
```

Reorder now means on-hand stock is at or below the reorder point. In the published sample that is BEARING, SEAL-KIT and HYDRAULIC-HOSE. SUPPORT is the high-stock row.
